```python
import sys, time, random, re
from pathlib import Path

# Add scripts dir to path so we can import scraper_utils
SCRIPTS_DIR = Path.home() / "Job_Scrapers" / "All_Scripts"
sys.path.insert(0, str(SCRIPTS_DIR))

from scraper_utils import *
import requests
from bs4 import BeautifulSoup
from datetime import datetime, timedelta

print("Imports loaded. Date:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
```

```python
COMPANY = "TCS"
OUTPUT_DIR = get_output_dir(COMPANY)
print(f"Output directory: {OUTPUT_DIR}")
```

```python
print("=" * 60)
print("TCS INDIA JOB SCRAPER")
print("Source: ibegin.tcs.com")
print("=" * 60)

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC


def fetch_jd_selenium(driver, url, timeout=10):
    """Visit a job detail page and extract the JD text."""
    try:
        driver.get(url)
        time.sleep(random.uniform(2, 4))
        soup = BeautifulSoup(driver.page_source, "lxml")
        # Try common JD container selectors
        for sel in ["[class*='job-description']", "[class*='jd-info']", "[class*='description']",
                     "[class*='details']", "article", "main", ".content"]:
            el = soup.select_one(sel)
            if el and len(el.get_text(strip=True)) > 100:
                return el.get_text(" ", strip=True)
        # Fallback: get body text
        body = soup.select_one("body")
        return body.get_text(" ", strip=True)[:5000] if body else ""
    except Exception as e:
        print(f"    [WARN] JD fetch failed for {url}: {e}")
        return ""

def fetch_jd_requests(session, url):
    """Fetch a job detail page via requests and extract JD text."""
    try:
        resp = session.get(url, timeout=20)
        if resp.status_code == 200:
            soup = BeautifulSoup(resp.text, "lxml")
            for sel in ["[class*='job-description']", "[class*='jd-info']", "[class*='description']",
                         "[class*='details']", "article", "main"]:
                el = soup.select_one(sel)
                if el and len(el.get_text(strip=True)) > 100:
                    return el.get_text(" ", strip=True)
            body = soup.select_one("body")
            return body.get_text(" ", strip=True)[:5000] if body else ""
    except:
        pass
    return ""


tcs_jobs = []

# Try the iBegin portal with Selenium (JS-heavy site)
driver = setup_selenium()
try:
    driver.get("https://ibegin.tcs.com/iBegin/jobs/search")
    time.sleep(10)

    # TCS may redirect or require interaction
    soup = BeautifulSoup(driver.page_source, "lxml")
    cards = soup.select("[class*='job-card'], [class*='job-listing'], [class*='search-result'], tr[class*='job']")
    if not cards:
        cards = soup.select("a[href*='/job/'], a[href*='/iBegin/']")

    for card in cards:
        title_el = card.select_one("h2, h3, h4, [class*='title'], td:first-child a")
        title = title_el.get_text(strip=True) if title_el else ""
        loc_el = card.select_one("[class*='location'], [class*='city']")
        loc = loc_el.get_text(strip=True) if loc_el else "India"

        if title and "india" in (loc + " " + card.get_text()).lower():
            link = card.select_one("a[href]")
            href = link.get("href", "") if link else ""
            tcs_jobs.append({
                "job_id": href.split("/")[-1] if href else str(len(tcs_jobs)),
                "title": title,
                "company_name": "TCS",
                "raw_jd_text": card.get_text(" ", strip=True),
                "location_city": loc.split(",")[0].strip(),
                "industry": "IT Services",
                "date_posted": datetime.now().strftime("%Y-%m-%d"),
                "is_active": True,
                "job_url": href if href.startswith("http") else f"https://ibegin.tcs.com{href}" if href else "",
                "business_unit": "",
                "source_platform": "Selenium",
            })

    print(f"  iBegin portal: {len(tcs_jobs)} jobs found")

except Exception as e:
    print(f"  iBegin error: {e}")
finally:
    driver.quit()

# Alternate: Try tcs.com/careers
if len(tcs_jobs) < 5:
    print("\n  Trying tcs.com careers...")
    driver = setup_selenium()
    try:
        driver.get("https://www.tcs.com/careers/india")
        time.sleep(8)
        soup = BeautifulSoup(driver.page_source, "lxml")
        cards = soup.select("[class*='job'], [class*='career'], a[href*='/careers/']")
        for card in cards:
            title = card.get_text(strip=True)[:100]
            if title and len(title) > 5:
                tcs_jobs.append({
                    "job_id": str(len(tcs_jobs)),
                    "title": title,
                    "company_name": "TCS",
                    "location_city": "India",
                    "industry": "IT Services",
                    "date_posted": datetime.now().strftime("%Y-%m-%d"),
                    "is_active": True,
                    "job_url": "",
                    "business_unit": "",
                    "source_platform": "Selenium",
                })
    except Exception as e:
        print(f"  TCS careers error: {e}")
    finally:
        driver.quit()

print(f"Total TCS India jobs: {len(tcs_jobs)}")
```

```python
df_tcs = save_results(tcs_jobs, "TCS", OUTPUT_DIR)
if df_tcs is not None:
    print(f"\nSample jobs:")
    cols = ["title","location_city","seniority_level","business_unit","job_url"]
    cols = [c for c in cols if c in df_tcs.columns]
    print(df_tcs[cols].head(10).to_string())
```
