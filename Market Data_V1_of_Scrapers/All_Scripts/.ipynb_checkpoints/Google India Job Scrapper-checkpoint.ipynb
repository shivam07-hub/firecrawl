```python
import sys, json, re, time, random
from pathlib import Path
from datetime import datetime
import xml.etree.ElementTree as ET

# Add scripts dir
SCRIPTS_DIR = Path.home() / 'Job_Scrapers' / 'All_Scripts'
sys.path.insert(0, str(SCRIPTS_DIR))

from scraper_utils import *
import requests
from bs4 import BeautifulSoup

print(f'[{datetime.now().strftime("%H:%M:%S")}] Imports loaded')
print(f'Schema columns: {len(SCHEMA_COLUMNS)}')

COMPANY = 'Google'
OUTPUT_DIR = get_output_dir(COMPANY)
print(f'Output dir: {OUTPUT_DIR}')
```

```python
# Helper functions - Sitemap + Selenium extraction
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

SITEMAP_URL = 'https://www.google.com/about/careers/applications/jobs/sitemap.xml'

def fetch_sitemap():
    """Fetch and parse XML sitemap"""
    print(f'\n[SITEMAP] Fetching: {SITEMAP_URL}')
    resp = requests.get(SITEMAP_URL, timeout=30)
    resp.raise_for_status()
    
    root = ET.fromstring(resp.content)
    ns = {'sm': 'http://www.sitemaps.org/schemas/sitemap/0.9'}
    
    urls = []
    for url_elem in root.findall('.//sm:url/sm:loc', ns):
        url = url_elem.text.strip()
        if '/jobs/results/' in url:
            urls.append(url)
    
    print(f'[SITEMAP] Found {len(urls)} job URLs')
    return urls

def extract_job_with_selenium(driver, url):
    """Use Selenium to load page and extract job details"""
    try:
        driver.get(url)
        # Wait for page to load - look for job title
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.TAG_NAME, "h1"))
        )
        time.sleep(2)  # Extra wait for dynamic content
        
        soup = BeautifulSoup(driver.page_source, 'lxml')
        
        # Extract job details from rendered page
        job_data = {}
        
        # Title - usually in h1
        title_elem = soup.find('h1')
        job_data['title'] = title_elem.get_text(strip=True) if title_elem else ''
        
        # Look for text containing locations
        # Google typically shows: "City, State, Country; +N more"
        page_text = soup.get_text()
        
        # Extract location - look for patterns
        loc_match = re.search(r'([A-Za-z\s]+,\s*[A-Z]{2},\s*[A-Z]{3,4})', page_text)
        if loc_match:
            location_str = loc_match.group(1)
            parts = [p.strip() for p in location_str.split(',')]
            job_data['location_city'] = parts[0] if len(parts) > 0 else ''
            job_data['location_state'] = parts[1] if len(parts) > 1 else ''
            job_data['location_country'] = parts[2] if len(parts) > 2 else ''
        else:
            job_data['location_city'] = ''
            job_data['location_state'] = ''
            job_data['location_country'] = ''
        
        # Get job description - look for "About the job" section or main content
        jd_section = soup.find(string=re.compile('About the job', re.I))
        if jd_section:
            jd_parent = jd_section.find_parent()
            if jd_parent:
                # Get next siblings or parent's text
                jd_text = jd_parent.get_text(' ', strip=True)[:5000]
                job_data['jd_text'] = jd_text
        else:
            # Fallback: just get main body text
            job_data['jd_text'] = page_text[:5000]
        
        return job_data
        
    except Exception as e:
        print(f'  [ERROR] Selenium extraction failed for {url}: {e}')
        return None

def parse_job_from_selenium(job_data, job_url, source_url):
    """Convert Selenium-extracted data to schema format"""
    if not job_data:
        return None
    
    # Extract job ID from URL
    match = re.search(r'/jobs/results/(\d+)', job_url)
    job_id = match.group(1) if match else ''
    
    return {
        'job_id': job_id,
        'title': job_data.get('title', ''),
        'location_city': job_data.get('location_city', ''),
        'location_state': job_data.get('location_state', ''),
        'location_country': job_data.get('location_country', ''),
        'jd_text': job_data.get('jd_text', ''),
        'job_url': job_url,
        'source_api_url': source_url,
        'company': 'Google',
    }

print('[READY] Selenium-based extraction functions loaded')
```

```python
# Main scraper execution with Selenium
print('='*70)
print('GOOGLE JOB SCRAPER v3.0 - SITEMAP + SELENIUM')
print('='*70)

# Step 1: Fetch sitemap
job_urls = fetch_sitemap()

# Step 2: Limit for testing (remove for production)
max_jobs = 5  # Testing with 5 jobs first
sample_urls = job_urls[:max_jobs] if max_jobs else job_urls

print(f'\n[FETCH] Processing {len(sample_urls)} jobs (out of {len(job_urls)} total)\n')

# Step 3: Initialize Selenium driver
driver = setup_chrome_driver()  # Uses scraper_utils function

try:
    google_jobs = []
    
    for i, job_url in enumerate(sample_urls, 1):
        print(f'[{i}/{len(sample_urls)}] Extracting: {job_url[:80]}...')
        
        job_data = extract_job_with_selenium(driver, job_url)
        if job_data:
            job = parse_job_from_selenium(job_data, job_url, SITEMAP_URL)
            if job:
                google_jobs.append(job)
                print(f'  ✓ {job["title"][:60]} | {job.get("location_city", "Unknown")}')
        
        # Respectful scraping
        time.sleep(random.uniform(2, 3))
    
    print(f'\n[COMPLETE] Successfully scraped {len(google_jobs)} jobs')
    
finally:
    driver.quit()
    print('[CLEANUP] Selenium driver closed')

# Show sample
if google_jobs:
    print('\n[SAMPLE] First 3 jobs:')
    for job in google_jobs[:3]:
        print(f'  - {job["title"]}')
        print(f'    Location: {job.get("location_city", "?")}, {job.get("location_country", "?")}')
        print(f'    URL: {job["job_url"][:70]}...')
```
