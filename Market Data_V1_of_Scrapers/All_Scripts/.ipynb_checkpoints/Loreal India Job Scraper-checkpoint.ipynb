# L'Oréal India Job Scraper

**Weekly Scraper - India Jobs Only**

**Source:** https://careers.loreal.com/en_US/jobs/SearchJobs

```python
# Configuration & Imports
import os, re, time, json
import pandas as pd
from datetime import datetime, timedelta
from pathlib import Path
import requests
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager

COMPANY = "Loreal"
BASE_DIR = Path.home() / "Job_Scrapers" / COMPANY
MONTH_TAG = datetime.now().strftime("%Y_%m")
OUTPUT_DIR = BASE_DIR / "Outputs" / MONTH_TAG
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f"Output directory: {OUTPUT_DIR}")
```

```python
# Helper functions
def extract_skills(text):
    """Extract tech/business skills from job text"""
    if not text: return []
    skills_db = ['python','java','sql','aws','azure','salesforce','sap','excel','tableau','power bi',
                 'marketing','digital','supply chain','finance','hr','legal','e-commerce','retail']
    t = text.lower()
    return [s for s in skills_db if re.search(r'\\b'+re.escape(s)+r'\\b', t)]

def parse_india_cities(text):
    """Detect Indian cities in location text"""
    india_cities = ['mumbai','delhi','bangalore','bengaluru','hyderabad','chennai','pune','kolkata',
                    'ahmedabad','gurgaon','noida','ghaziabad','jaipur','kochi','chandigarh']
    if not text: return None
    t = text.lower()
    for city in india_cities:
        if city in t:
            return city.title()
    return None

def setup_driver():
    """Setup headless Chrome"""
    opts = Options()
    opts.add_argument('--headless')
    opts.add_argument('--no-sandbox')
    opts.add_argument('--disable-dev-shm-usage')
    opts.add_argument('user-agent=Mozilla/5.0 (Macintosh; Intel Mac OS X 10_15_7)')
    service = Service(ChromeDriverManager().install())
    return webdriver.Chrome(service=service, options=opts)

COLS = ['job_id','title','company_name','raw_jd_text','skills_required','skills_preferred',
        'min_years_experience','max_years_experience','seniority_level',
        'location_city','location_country','work_mode','employment_type',
        'degree_required','degree_preferred_field','industry','salary_min','salary_max','salary_currency',
        'date_posted','is_active','division_or_brand','function_area','job_url','scraped_at']

print('Helper functions loaded')
```

```python
# Main Scraper Function
def scrape_loreal_india():
    """Scrape all India L'Oreal jobs using Selenium to filter for India"""
    jobs = []
    driver = setup_driver()
    
    try:
        # Navigate to jobs search page
        url = 'https://careers.loreal.com/en_US/jobs/SearchJobs'
        print(f'Opening {url}...')
        driver.get(url)
        time.sleep(5)  # Let page load
        
        # Select India from Location filter
        try:
            # Click Location dropdown
            loc_dropdown = WebDriverWait(driver, 10).until(
                EC.element_to_be_clickable((By.XPATH, "//select[@aria-label='Location']|//button[contains(@aria-label,'Location')]|//div[contains(text(),'Location')]"))
            )
            loc_dropdown.click()
            time.sleep(2)
            
            # Type 'India' to filter
            search_box = driver.find_element(By.CSS_SELECTOR, ".select2-search__field")
            search_box.send_keys('India')
            time.sleep(1)
            
            # Click India option
            india_option = WebDriverWait(driver, 5).until(
                EC.element_to_be_clickable((By.XPATH, "//li[contains(text(),'India')]|//option[contains(text(),'India')]"))
            )
            india_option.click()
            time.sleep(2)
            
            # Click Search button
            search_btn = driver.find_element(By.XPATH, "//button[contains(text(),'Search')]|//button[@type='submit']")
            search_btn.click()
            time.sleep(5)
        except Exception as e:
            print(f'Location filter error: {e}. Will parse all jobs and filter by location text.')
        
        # Scrape jobs from current page view
        page = 1
        while page <= 10:  # Limit to 10 pages
            print(f'\nScraping page {page}...')
            
            # Parse job listings
            soup = BeautifulSoup(driver.page_source, 'html.parser')
            job_cards = soup.find_all('article', class_=lambda x: x and 'job' in x.lower()) or soup.find_all('div', class_=lambda x: x and 'job' in x.lower())
            
            if not job_cards:
                # Try alternative selectors
                job_cards = soup.find_all('a', href=lambda x: x and 'JobDetail' in x)
            
            print(f'Found {len(job_cards)} job cards on page {page}')
            
            for card in job_cards:
                try:
                    # Extract job URL
                    job_link = card.get('href') or card.find('a', href=lambda x: x and 'JobDetail' in x)
                    if isinstance(job_link, str):
                        job_url = f"https://careers.loreal.com{job_link}" if job_link.startswith('/') else job_link
                    else:
                        job_url = f"https://careers.loreal.com{job_link.get('href')}" if job_link else ''
                    
                    # Extract title
                    title = ''
                    title_tag = card.find('h3') or card.find('h2') or card.find('a')
                    if title_tag:
                        title = title_tag.get_text(strip=True)
                    
                    # Extract location
                    location = ''
                    loc_tag = card.find(string=lambda x: x and any(c in x.lower() for c in ['mumbai','delhi','bangalore','hyderabad','chennai','pune','india']))
                    if loc_tag:
                        location = loc_tag.strip()
                    else:
                        # Try all text
                        location = card.get_text(' ', strip=True)
                    
                    # Filter for India jobs
                    india_city = parse_india_cities(location)
                    if not india_city and 'india' not in location.lower():
                        continue  # Skip non-India jobs
                    
                    # Extract date posted
                    date_posted = ''
                    date_tag = card.find(string=lambda x: x and 'posted' in x.lower())
                    if date_tag:
                        date_posted = date_tag.strip()
                    
                    # Extract job ID from URL
                    job_id = job_url.split('/')[-1] if job_url else f"loreal_{len(jobs)}"
                    
                    # Scrape job detail page for full JD
                    raw_jd = ''
                    division = ''
                    function_area = ''
                    if job_url:
                        try:
                            driver.get(job_url)
                            time.sleep(3)
                            detail_soup = BeautifulSoup(driver.page_source, 'html.parser')
                            
                            # Get job description
                            jd_div = detail_soup.find('div', class_=lambda x: x and 'description' in x.lower()) or detail_soup.find('div', id=lambda x: x and 'description' in x.lower())
                            if jd_div:
                                raw_jd = jd_div.get_text('\n', strip=True)
                            
                            # Extract division/brand from meta fields
                            for meta in detail_soup.find_all(['span','div','dd']):
                                txt = meta.get_text(strip=True)
                                if any(b in txt.lower() for b in ['luxe','consumer','professional','dermatological']):
                                    division = txt
                                if any(f in txt.lower() for f in ['marketing','sales','supply','hr','finance','it','legal']):
                                    function_area = txt
                            
                            driver.back()
                            time.sleep(2)
                        except:
                            pass
                    
                    # Build job dict
                    job = {
                        'job_id': job_id,
                        'title': title,
                        'company_name': 'L\'Oréal',
                        'raw_jd_text': raw_jd,
                        'skills_required': ','.join(extract_skills(raw_jd)),
                        'location_city': india_city or location.split(',')[0].strip() if location else '',
                        'location_country': 'India',
                        'industry': 'Beauty & Cosmetics',
                        'division_or_brand': division,
                        'function_area': function_area,
                        'date_posted': date_posted,
                        'is_active': True,
                        'job_url': job_url,
                        'scraped_at': datetime.now().isoformat(),
                        'salary_currency': 'INR'
                    }
                    
                    jobs.append(job)
                    print(f'  ✓ {title} - {job.get("location_city","")}')
                    
                except Exception as e:
                    print(f'  ✗ Card parse error: {e}')
                    continue
            
            # Try to go to next page
            try:
                next_btn = driver.find_element(By.XPATH, "//a[contains(text(),'Next')]|//button[contains(text(),'Next')]|//a[@aria-label='Next']")
                if 'disabled' in next_btn.get_attribute('class') or not next_btn.is_enabled():
                    print('No more pages.')
                    break
                next_btn.click()
                time.sleep(4)
                page += 1
            except:
                print('No next button found.')
                break
    
    finally:
        driver.quit()
    
    print(f'\nTotal India jobs scraped: {len(jobs)}')
    return jobs

print('Scraper function ready')
```

```python
# Execute Scraper
print(f'Starting L\'Oréal India job scraper at {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}\n')
start_time = time.time()

# Run scraper
jobs_data = scrape_loreal_india()

# Create DataFrame
df = pd.DataFrame(jobs_data, columns=COLS)
df = df.fillna('')

# Display summary
print(f'\n{"=" * 60}')
print(f'Scraping completed in {time.time() - start_time:.1f} seconds')
print(f'Total jobs found: {len(df)}')
if len(df) > 0:
    print(f'Cities: {df.location_city.value_counts().to_dict()}')
    print(f'Divisions: {df.division_or_brand.value_counts().head(5).to_dict()}')
print(f'{"="*60}\n')

# Save to CSV
output_file = OUTPUT_DIR / f"Loreal_India_Jobs_{datetime.now().strftime('%Y%m%d')}.csv"
df.to_csv(output_file, index=False)
print(f'Saved {len(df)} jobs to: {output_file}')
df.head(10)
```
