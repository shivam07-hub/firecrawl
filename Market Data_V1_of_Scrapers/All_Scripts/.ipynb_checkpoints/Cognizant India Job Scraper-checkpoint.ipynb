```python
import sys, time, random, re
from pathlib import Path

# Add scripts dir to path so we can import scraper_utils
SCRIPTS_DIR = Path.home() / "Job_Scrapers" / "All_Scripts"
sys.path.insert(0, str(SCRIPTS_DIR))

from scraper_utils import *
import requests
from bs4 import BeautifulSoup
from datetime import datetime, timedelta

print("Imports loaded. Date:", datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
```

```python
COMPANY = "Cognizant"
OUTPUT_DIR = get_output_dir(COMPANY)
print(f"Output directory: {OUTPUT_DIR}")
```

```python
print("=" * 60)
print("COGNIZANT INDIA JOB SCRAPER")
print("Source: careers.cognizant.com/india-en/jobs (RSS + Selenium)")
print("=" * 60)

cognizant_jobs = []
session = get_session()

# Try RSS/XML feed first (most reliable)
try:
    rss_url = "https://careers.cognizant.com/india-en/jobs/xml/?rss=true"
    print(f"  Trying RSS feed: {rss_url}")
    resp = session.get(rss_url, timeout=20)
    if resp.status_code == 200 and ("<rss" in resp.text[:200] or "<item" in resp.text[:500]):
        soup = BeautifulSoup(resp.text, "xml")
        items = soup.find_all("item")
        for item in items:
            title = item.find("title").get_text(strip=True) if item.find("title") else ""
            link = item.find("link").get_text(strip=True) if item.find("link") else ""
            desc = item.find("description").get_text(strip=True) if item.find("description") else ""
            pub_date = item.find("pubDate").get_text(strip=True) if item.find("pubDate") else ""

            loc = "India"
            loc_el = item.find("location")
            if loc_el:
                loc = loc_el.get_text(strip=True)

            if title:
                cognizant_jobs.append({
                    "job_id": link.split("/")[-1] if link else str(len(cognizant_jobs)),
                    "title": title,
                    "company_name": "Cognizant",
                    "raw_jd_text": html_to_text(desc),
                    "location_city": loc.split(",")[0].strip(),
                    "industry": "IT Services & Consulting",
                    "date_posted": pub_date[:10] if pub_date else datetime.now().strftime("%Y-%m-%d"),
                    "is_active": True,
                    "job_url": link,
                    "business_unit": "",
                    "source_platform": "Cognizant Selenium fallback",
                })
        print(f"  RSS feed: {len(cognizant_jobs)} jobs")
    else:
        print(f"  RSS returned status {resp.status_code} or non-RSS content")
except Exception as e:
    print(f"  RSS failed: {e}")

# Selenium fallback
if len(cognizant_jobs) < 5:
    print("  Trying Selenium on careers.cognizant.com...")
    from selenium.webdriver.common.by import By
    driver = setup_selenium()
    try:
        driver.get("https://careers.cognizant.com/india-en/jobs/")
        time.sleep(10)
        soup = BeautifulSoup(driver.page_source, "lxml")
        cards = soup.select("[class*='job-card'], [class*='job-listing'], a[href*='/job/'], [class*='search-result']")

        for card in cards:
            title_el = card.select_one("h2, h3, h4, [class*='title']") or card
            title = title_el.get_text(strip=True)
            loc_el = card.select_one("[class*='location']")
            loc = loc_el.get_text(strip=True) if loc_el else "India"

            if title and len(title) > 5 and title not in [j["title"] for j in cognizant_jobs]:
                cognizant_jobs.append({
                    "job_id": str(len(cognizant_jobs)),
                    "title": title,
                    "company_name": "Cognizant",
                    "raw_jd_text": card.get_text(" ", strip=True),
                    "location_city": loc.split(",")[0].strip(),
                    "industry": "IT Services & Consulting",
                    "date_posted": datetime.now().strftime("%Y-%m-%d"),
                    "is_active": True,
                    "job_url": "",
                    "business_unit": "",
                    "source_platform": "Cognizant Selenium fallback",
                })
    except Exception as e:
        print(f"  Selenium error: {e}")
    finally:
        driver.quit()

print(f"Total Cognizant India jobs: {len(cognizant_jobs)}")
```

```python
df_cognizant = save_results(cognizant_jobs, "Cognizant", OUTPUT_DIR)
if df_cognizant is not None:
    print(f"\nSample jobs:")
    cols = ["title","location_city","seniority_level","business_unit","job_url"]
    cols = [c for c in cols if c in df_cognizant.columns]
    print(df_cognizant[cols].head(10).to_string())
```
